## Step 6: Launch Phase 2 Training (YOLO11 Medium - 20.1M Parameters)
With a 16GB T4 GPU, we scale batch size to **32** at **640px** (or 800px).

```python
from ultralytics import YOLO

# Initialize YOLO11 Medium (20.1M parameters - 7.5x larger than Nano)
model = YOLO('yolo11m.pt')

# Train with auto-checkpointing to Google Drive
model.train(
    data='data/yolo_dataset/dataset.yaml',
    epochs=5,
    batch=32,
    imgsz=640,
    device=0,
    project='/content/drive/MyDrive/CV_Project_Phase2/runs',
    name='phase2_epochs_1_to_5',
    save=True,
    save_period=1,
    plots=True
)
```

## Step 7: Test the High-Capacity Model on Difficult Images
Test inference on real-world scenes (including images with distant cars and dense crowds).

```python
from ultralytics import YOLO
from PIL import Image
import glob

# Load best Phase 2 trained weights
best_weights = '/content/drive/MyDrive/CV_Project_Phase2/runs/phase2_epochs_1_to_5/weights/best.pt'
eval_model = YOLO(best_weights)

# Test on images3.jfif or any custom uploaded image
results = eval_model.predict(source='images3.jfif', conf=0.25, save=True)

# Display result
saved_img = glob.glob(results[0].save_dir + '/*')[0]
display(Image.open(saved_img))
print(f'Detected {len(results[0].boxes)} objects with high confidence!')
```
