## Step 6.1: Round 1 (Epochs 1 to 5)
- Trains for 5 epochs.
- Saves weights to `phase2_epochs_1_to_5`.
- Automatically generates Report (.docx), LinkedIn Graphic (.png), and Post (.md) at completion.

```python
# Move to project directory and ensure dependencies are installed
%cd /content/CV_Project
!pip install -q ultralytics python-docx

import re
# Ensure dataset.yaml points to Colab Linux path
yaml_path = '/content/CV_Project/data/yolo_dataset/dataset.yaml'
with open(yaml_path, 'r', encoding='utf-8') as f:
    content = f.read()
content = re.sub(r'path: .*', 'path: /content/CV_Project/data/yolo_dataset', content)
with open(yaml_path, 'w', encoding='utf-8') as f:
    f.write(content)

from ultralytics import YOLO

# 1. Launch Round 1: Epochs 1 to 5
model = YOLO('yolo11m.pt')
results = model.train(
    data='/content/CV_Project/data/yolo_dataset/dataset.yaml',
    epochs=5,
    batch=32,
    imgsz=640,
    device=0,
    project='/content/drive/MyDrive/CV_Project_Phase2/runs',
    name='phase2_epochs_1_to_5',
    save=True,
    save_period=1,
    plots=True
)

# 2. Auto-generate Report, LinkedIn Graphic & Post for 5 Epochs
print('\nGenerating Phase 2 Round 1 (5 Epochs) Deliverables...')
!python scripts/generate_phase2_artifacts.py --round 1 --epochs 5 --results_csv '/content/drive/MyDrive/CV_Project_Phase2/runs/phase2_epochs_1_to_5/results.csv' --output_dir '/content/drive/MyDrive/CV_Project_Phase2'
print('DONE: Round 1 deliverables saved in Google Drive (Reports, Posts, Documentation)!')
```
